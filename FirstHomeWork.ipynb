## 1 Задание: Мини-скрипт с обоснованным выбором коллекций
Для первого задания я выбрал трекер рыб в аквариуме. В следующей ячейке предоставлен код.

```python
# Список всех рыб аквариума.
# Каждая рыба представлена tuple: (id, имя, вид).
fish_list = [
    (1, "Nemo", "clownfish"),
    (2, "Dory", "blue tang"),
    (3, "Bubbles", "clownfish"),
]

# Словарь для быстрого поиска рыбы по уникальному ID.
fish_by_id = {
    fish[0]: fish
    for fish in fish_list
}

# Множество всех видов, встречающихся в аквариуме.
species = {
    fish[2]
    for fish in fish_list
}
```

- `list` используется для хранения текущего набора рыб, потому что рыбы могут добавляться и удаляться из аквариума.
- `tuple` используется для отдельной записи о рыбе, так как её структура фиксирована: ID, имя, вид.
- `dict` хранит записи по уникальному ID и позволяет быстро находить конкретную рыбу.
- `set` используется для хранения видов рыб, поскольку один вид может встречаться у нескольких рыб, но в этом случае нам нужны только уникальные названия видов.

Добавим новую рыбу

```python
# Добавляем новую рыбу.
new_fish = (4, "Sunny", "goldfish", 1)

fish_list.append(new_fish)
fish_by_id[new_fish[0]] = new_fish
species.add(new_fish[2])
```

Выведем список всех рыб в аквариуме, найдём рыбу по ID, и найдем все уникальные виды

```python
print("Рыбы в аквариуме:")
for fish in fish_list:
    print(fish)
print('---')
print("Рыба с ID 2:")
print(fish_by_id[2])
print('---')
print("Уникальные виды:")
print(species)
```

## 2 задание: Свой баг-репорт

```python
# Багованный код
aquarium = {
    "Nemo": "clownfish",
    "Dory": "blue tang",
    "Bubbles": "goldfish"
}

for name, species in aquarium.items():
    if species == "goldfish":
        del aquarium[name]

print(aquarium)
```

Ошибка возникает потому, что словарь нельзя изменять по размеру во время обхода его элементов. Удаление записи внутри цикла нарушает текущую итерацию.

```python
aquarium = {
    "Nemo": "clownfish",
    "Dory": "blue tang",
    "Bubbles": "goldfish"
}

for name, species in list(aquarium.items()):
    if species == "goldfish":
        del aquarium[name]

print(aquarium)
```
